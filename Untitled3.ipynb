```python
import random

def calculate_attacks(state):
    """Calculates the number of attacking queen pairs (heuristic value h)
    where state[row] = col."""
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # i and j represent rows. state[i] and state[j] represent columns.
            # Check if they are in the same column or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_best_neighbor(state):
    """Finds the neighbor with the lowest heuristic value (moving queens column-wise per row)."""
    n = len(state)
    current_h = calculate_attacks(state)
    best_neighbors = []
    best_h = current_h

    # Iterate through rows and test moving the queen to different columns
    for row in range(n):
        for col in range(n):
            if state[row] == col:
                continue
            neighbor = list(state)
            neighbor[row] = col
            h = calculate_attacks(neighbor)

            if h < best_h:
                best_h = h
                best_neighbors = [neighbor]
            elif h == best_h:
                best_neighbors.append(neighbor)

    if best_neighbors:
        return random.choice(best_neighbors), best_h
    return state, current_h

def print_board(state, h_value):
    """Prints the board visual using row-indexed logic."""
    n = len(state)
    print("-" * (n * 3))
    for row in range(n):
        line = ""
        for col in range(n):
            # Check if the queen for this row is in this column
            if state[row] == col:
                line += " Q "
            else:
                line += " . "
        print(line)
    print(f"State: {state} | Heuristic (h) = {h_value}")
    print("-" * (n * 3))

def hill_climbing_search(n, max_restarts=100):
    print(f"Solving {n}-Queens using Hill Climbing (Row-indexed State)...\n")

    for attempt in range(1, max_restarts + 1):
        print(f"\n=== ATTEMPT #{attempt} ===")
        # Assign queens randomly column-wise for each row
        # State index = Row, State value = Column
        state = [random.randint(0, n - 1) for _ in range(n)]
        current_h = calculate_attacks(state)

        step = 1
        while True:
            print(f"\nStep {step}:")
            print_board(state, current_h)

            if current_h == 0:
                print("\n🎉 --- GOAL STATE FOUND --- 🎉")
                return state

            neighbor, neighbor_h = get_best_neighbor(state)

            if neighbor_h >= current_h:
                print(f"Stuck at local minimum (h={current_h}). Restarting...")
                break

            state = neighbor
            current_h = neighbor_h
            step += 1

    print(f"\nCould not find a solution within {max_restarts} restarts.")
    return None

# Execution
if __name__ == "__main__":
    Queens = int(input("Enter the number of Queens = "))
    solution = hill_climbing_search(Queens)
```
