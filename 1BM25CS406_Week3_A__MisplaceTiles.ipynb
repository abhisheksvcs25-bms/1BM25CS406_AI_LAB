```python
import heapq

# Define the global goal state (0 represents the blank tile)
GOAL_STATE = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

# Valid movement shifts for the blank tile (row_offset, col_offset, label)
MOVES = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]

def get_neighbors(state):
    """Finds all legal board layouts achievable by moving the blank space."""
    neighbors = []
    blank_idx = state.index(0)
    r, c = divmod(blank_idx, 3)

    for dr, dc, move_name in MOVES:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            neighbor_idx = nr * 3 + nc
            new_state = list(state)
            # Swap the blank tile with the target tile
            new_state[blank_idx], new_state[neighbor_idx] = new_state[neighbor_idx], new_state[blank_idx]
            neighbors.append((tuple(new_state), move_name))

    return neighbors

def h_manhattan_distance(state):
    """Heuristic function: Calculates sum of grid steps needed for tiles to reach their goal."""
    distance = 0
    for i in range(9):
        val = state[i]
        # Skip calculating layout offset distance for the blank tile (0)
        if val != 0:
            current_row, current_col = divmod(i, 3)
            goal_idx = GOAL_STATE.index(val)
            goal_row, goal_col = divmod(goal_idx, 3)
            distance += abs(current_row - goal_row) + abs(current_col - goal_col)
    return distance

def a_star_manhattan(start_state):
    """Executes A* Search using Manhattan Distance heuristic."""
    # Priority Queue elements format: (f_score, g_score, current_layout, step_history)
    h_start = h_manhattan_distance(start_state)
    pq = [(h_start, 0, start_state, [])]

    # Tracking visited states to optimize and prevent cycles
    visited = {start_state: 0}

    while pq:
        f, g, current, path = heapq.heappop(pq)

        if current == GOAL_STATE:
            return path, g

        for neighbor, move in get_neighbors(current):
            new_g = g + 1
            if neighbor not in visited or new_g < visited[neighbor]:
                visited[neighbor] = new_g
                h = h_manhattan_distance(neighbor)
                new_f = new_g + h
                heapq.heappush(pq, (new_f, new_g, neighbor, path + [move]))

    return None, -1

if __name__ == "__main__":
    # Supply your initial layout configuration here
    initial_state = (1, 2, 3,
                     0, 4, 6,
                     7, 5, 8)

    print("--- A* Search: Manhattan Distance Heuristic ---")
    print(f"Initial Configuration: {initial_state}")

    solution_path, total_depth = a_star_manhattan(initial_state)

    if solution_path is not None:
        print(f"Goal reached at depth (g): {total_depth}")
        print(f"Move Sequence: {solution_path}")
    else:
        print("No viable solution sequence found.")
```
